```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
```

### Loading the data

```python
data = pd.read_csv("Downloads/Motor vehicle insurance data.csv", delimiter = ';')
data.head()
```

### Cleaning the data

```python
data.info()
```

```python
data.describe()
```

```python
data.columns
```

```python
data.isnull().sum()
```

```python
data['Date_lapse'].sample(5)
```

```python
data['Policy_active'] = data['Date_lapse'].isnull().astype(int) # Assumed that null values meant that the policy is active
```

```python
data['Policy_active'].sample(10)
```

```python
# Customer age
data['Customer_age'] = pd.to_datetime('today').year - pd.to_datetime(data['Date_birth'], dayfirst=True).dt.year
```

```python
# Car age
data['Car_age'] = pd.to_datetime('today').year - data['Year_matriculation']
```

```python
data[['Customer_age','Car_age']].sample(10)
```

```python
data.head()
```

```python
data.drop(columns = ["Poloicy_active"], inplace = True)
```

```python
data['Type_fuel'].value_counts(dropna=False)
```

```python
data["Type_fuel"].sample(5) # could transfrom this column into a binary column.
```

```python
data['Type_fuel'].fillna(data['Type_fuel'].mode()[0], inplace=True) # filled the mode/ most common fuel type to replace the missing values
```

```python
data['Type_fuel'].value_counts(dropna=False)
```

```python
data['Date_lapse'].fillna('Active') #This would indicate whether policy is active or ended.
```

```python
data['Length'].describe()
```

```python
data['Length'].fillna(data['Length'].median()) 
#Filled in the median length of cars. This will not affect outliers.
```
